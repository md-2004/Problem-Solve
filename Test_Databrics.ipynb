```python
from pyspark.sql import functions as F

path = "/Volumes/workspace/default/sample/erroneous_dataset.csv"

df = spark.read.csv(path, header=True, inferSchema=True)
```

```python
print("Rows:", df.count())
print("Columns:", len(df.columns))

display(df.limit(10))     # first 10 rows
   
```

```python
print("Rows:", df.count())
print("Columns:", len(df.columns))

df.printSchema() 
```

```python
total = df.count()

missing = df.select([
    F.sum(F.when(F.col(c).isNull() | (F.trim(F.col(c).cast("string")) == ""), 1).otherwise(0)).alias(c)
    for c in df.columns
])

# turn it into one row per column
missing_long = missing.selectExpr(
    "stack({}, {}) as (column_name, missing_count)".format(
        len(df.columns), ", ".join([f"'{c}', `{c}`" for c in df.columns])
    )
).withColumn("missing_pct", F.round(100 * F.col("missing_count") / total, 2))

display(missing_long.orderBy(F.desc("missing_count")))
```

```python
# Description_Payload is just random text, so we remove it
df = df.drop("Description_Payload")

print("Columns now:", len(df.columns))
```

```python
# trim() removes extra spaces, so " Finance " becomes "Finance"
text_columns = ["Employee_ID", "Employee_Name", "Department", "Designation",
                "City", "Employment_Status", "Email", "Manager_ID"]

for col in text_columns:
    df = df.withColumn(col, F.trim(F.col(col)))

# make email lowercase
df = df.withColumn("Email", F.lower(F.col("Email")))

# change the date columns from text to date
df = df.withColumn("Joining_Date", F.to_date("Joining_Date"))
df = df.withColumn("Last_Appraisal", F.to_date("Last_Appraisal"))

display(df.limit(5))
```

```python
total = df.count()

problems = [
    ("Missing name",            df.filter(F.col("Employee_Name").isNull()).count()),
    ("Missing department",      df.filter(F.col("Department").isNull()).count()),
    ("Wrong age",               df.filter(~F.col("Age").between(18, 65)).count()),
    ("Negative salary",         df.filter(F.col("Annual_Salary") <= 0).count()),
    ("Negative bonus",          df.filter(F.col("Annual_Bonus") < 0).count()),
    ("Wrong performance score", df.filter(~F.col("Performance_Score").between(0, 5)).count()),
    ("Wrong working hours",     df.filter(~F.col("Working_Hours").between(0, 16)).count()),
    ("Experience too high",     df.filter(F.col("Experience_Years") > F.col("Age") - 18).count()),
    ("Wrong email",             df.filter(F.col("Email").contains("@@")).count()),
    ("Duplicate employee IDs",  total - df.select("Employee_ID").distinct().count()),
]

dq = spark.createDataFrame(problems, ["problem", "rows"])
dq.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("data_quality_report")
display(dq)
```

```python
from pyspark.sql.window import Window
from pyspark.sql import functions as F

# the real employee is the one whose email number matches the ID number
df = df.withColumn("match", F.col("Email").contains(
        F.concat(F.expr("cast(cast(substring(Employee_ID, 4) AS INT) AS STRING)"), F.lit("@"))).cast("int"))

w = Window.partitionBy("Employee_ID").orderBy(F.desc("match"))
df = df.withColumn("n", F.row_number().over(w)).filter("n = 1").drop("match", "n")

print("Rows after removing duplicates:", df.count())   # 996,340
```

```python
# Experience goes first, because it is checked against the original Age
df = df.withColumn("Experience_Years", F.when(F.col("Experience_Years") <= F.col("Age") - 18, F.col("Experience_Years")))
df = df.withColumn("Age",               F.when(F.col("Age").between(18, 65), F.col("Age")))
df = df.withColumn("Annual_Salary",     F.when(F.col("Annual_Salary") > 0, F.col("Annual_Salary")))
df = df.withColumn("Annual_Bonus",      F.when(F.col("Annual_Bonus") >= 0, F.col("Annual_Bonus")))
df = df.withColumn("Performance_Score", F.when(F.col("Performance_Score").between(0, 5), F.col("Performance_Score")))
df = df.withColumn("Working_Hours",     F.when(F.col("Working_Hours").between(0, 16), F.col("Working_Hours")))
df = df.withColumn("Email",             F.when(~F.col("Email").contains("@@"), F.col("Email")))
```

```python
df = df.fillna({"Department": "Unknown", "Employee_Name": "Unknown"})
df = df.withColumn("Has_Left", F.when(F.col("Employment_Status") == "Resigned", 1).otherwise(0))
```

```python
df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("clean_employees")
print("Clean rows:", df.count())
display(df.limit(5))
```

```python
emp = spark.table("clean_employees")

report = emp.groupBy("Department").agg(
    F.count("*").alias("people"),
    F.sum("Has_Left").alias("people_left"),
    F.round(F.avg("Annual_Salary"), 0).alias("avg_salary"),
    F.round(F.sum("Annual_Salary"), 0).alias("total_salary"),
    F.round(F.avg("Performance_Score"), 2).alias("avg_performance"),
)

report.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("department_report")
display(report)
```

```python
reporting = emp.select("Manager_ID", "Employee_ID", "Employee_Name", "Department")
reporting.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("reporting_structure")

display(reporting.groupBy("Manager_ID").count())                   # team size per manager
display(reporting.filter(F.col("Manager_ID") == "MGR000002"))      # one manager's team
```
